## Dot Product — Geometric Interpretation

**Definition:** For $\mathbf{a}, \mathbf{b} \in \mathbb{R}^n$, the dot product has the geometric form:

$$\mathbf{a} \cdot \mathbf{b} = \|\mathbf{a}\|\,\|\mathbf{b}\|\cos\theta$$

where $\theta \in [0, \pi]$ is the angle between the vectors.

**Angle cases:**

| $\theta$ | $\cos\theta$ | $\mathbf{a}\cdot\mathbf{b}$ | Interpretation |
|---|---|---|---|
| $0$ | $1$ | $\|\mathbf{a}\|\|\mathbf{b}\|$ | Parallel |
| $(0, \pi/2)$ | $> 0$ | $> 0$ | Acute |
| $\pi/2$ | $0$ | $0$ | **Orthogonal** |
| $(\pi/2, \pi)$ | $< 0$ | $< 0$ | Obtuse |
| $\pi$ | $-1$ | $-\|\mathbf{a}\|\|\mathbf{b}\|$ | Anti-parallel |

**Projection:** The vector projection of $\mathbf{b}$ onto $\mathbf{a}$:

$$\text{proj}_{\mathbf{a}}\mathbf{b} = \frac{\mathbf{a}\cdot\mathbf{b}}{\|\mathbf{a}\|^2}\,\mathbf{a}$$

with scalar magnitude $\|\mathbf{b}\|\cos\theta$.


```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([3.0, 1.0])
b = np.array([1.0, 2.5])

dot = np.dot(a, b)
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)
theta = np.arccos(dot / (norm_a * norm_b))
proj_vec = (dot / norm_a**2) * a

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax in axes:
    ax.set_xlim(-0.3, 4.2)
    ax.set_ylim(-0.3, 3.2)
    ax.set_aspect("equal")
    ax.axhline(0, color="white", linewidth=0.5)
    ax.axvline(0, color="white", linewidth=0.5)
    ax.grid(alpha=0.2)
    for vec, label, color in [
        (a, r"$\mathbf{a}$", "orange"),
        (b, r"$\mathbf{b}$", "deepskyblue"),
    ]:
        ax.annotate(
            "",
            xy=vec,
            xytext=(0, 0),
            arrowprops=dict(arrowstyle="->", color=color, lw=2),
        )
        ax.text(vec[0] * 0.52, vec[1] * 0.52, label, fontsize=13, color=color)

# ── Left: angle ──────────────────────────────────────────────────────────────
ax = axes[0]
arc_t = np.linspace(0, theta, 60)
r = 0.55
ax.plot(r * np.cos(arc_t), r * np.sin(arc_t), color="white", lw=1.2)
mid = (r + 0.18) * np.array([np.cos(theta / 2), np.sin(theta / 2)])
ax.text(mid[0], mid[1], r"$\theta$", color="white", fontsize=12)
ax.set_title(
    rf"$\mathbf{{a}}\cdot\mathbf{{b}} = \|\mathbf{{a}}\|\|\mathbf{{b}}\|\cos\theta"
    rf" = {dot:.2f}$   $(\theta={np.degrees(theta):.1f}°)$",
    fontsize=10,
)

# ── Right: projection ─────────────────────────────────────────────────────────
ax = axes[1]
ax.annotate(
    "",
    xy=proj_vec,
    xytext=(0, 0),
    arrowprops=dict(arrowstyle="->", color="limegreen", lw=2),
)
ax.text(
    proj_vec[0] * 0.48,
    proj_vec[1] * 0.48 - 0.18,
    r"$\mathrm{proj}_{\mathbf{a}}\mathbf{b}$",
    color="limegreen",
    fontsize=11,
)
ax.plot([b[0], proj_vec[0]], [b[1], proj_vec[1]], "--", color="white", lw=1, alpha=0.5)

# right-angle marker
perp = b - proj_vec
perp_u = perp / np.linalg.norm(perp) * 0.15
a_u = a / norm_a * 0.15
corner = proj_vec + perp_u + a_u
ax.plot(
    [proj_vec[0] + perp_u[0], corner[0], proj_vec[0] + a_u[0]],
    [proj_vec[1] + perp_u[1], corner[1], proj_vec[1] + a_u[1]],
    color="white",
    lw=0.8,
)

ax.set_title(
    r"$\mathrm{proj}_{\mathbf{a}}\mathbf{b}"
    r"= \dfrac{\mathbf{a}\cdot\mathbf{b}}{\|\mathbf{a}\|^2}\,\mathbf{a}$",
    fontsize=11,
)

plt.suptitle(
    "Geometric Interpretation of the Dot Product", fontsize=13, fontweight="bold"
)
plt.tight_layout()
plt.show()
```

## Matrix Multiplication as a Grid of Dot Products

**Definition:** For $A \in \mathbb{R}^{m \times n}$, $B \in \mathbb{R}^{n \times p}$, the product $C = AB \in \mathbb{R}^{m \times p}$ is defined entry-wise as:

$$C_{ij} = \mathbf{a}_i \cdot \mathbf{b}_j = \sum_{k=1}^n A_{ik}\, B_{kj}$$

where $\mathbf{a}_i$ is the $i$-th **row** of $A$ and $\mathbf{b}_j$ is the $j$-th **column** of $B$. Every entry of $C$ is exactly a dot product — so $C$ is an $m \times p$ table of dot products.

The shared inner dimension $n$ is the one being summed over (contracted). Geometrically, $C_{ij}$ measures how much row $i$ of $A$ aligns with column $j$ of $B$, scaled by their magnitudes.

**Three equivalent views:**

| View | Expression | Intuition |
|---|---|---|
| Entry-wise | $C_{ij} = \mathbf{a}_i \cdot \mathbf{b}_j$ | Grid of dot products |
| Column-wise | $\mathbf{c}_j = A\mathbf{b}_j$ | $A$ transforms each column of $B$ |
| Outer product sum | $C = \sum_{k=1}^n \mathbf{a}^{(k)}(\mathbf{b}^{(k)})^T$ | Sum of rank-1 matrices |


```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# ------------------ CONFIG ------------------
SEED = 7
A_SHAPE = (3, 4)
B_SHAPE = (4, 3)
HIGHLIGHT_POS = (1, 2)  # (row, col)

CMAP = "plasma"
TEXT_COLOR = "white"

# ------------------ SETUP ------------------
np.random.seed(SEED)

def make_matrix(shape, low=1, high=7):
    return np.random.randint(low, high, size=shape).astype(float)

A = make_matrix(A_SHAPE)
B = make_matrix(B_SHAPE)
C = A @ B

hi_row, hi_col = HIGHLIGHT_POS

# ------------------ DRAW HELPERS ------------------
def add_highlight(ax, xy, width, height, color, alpha=0.2):
    """Draw filled + border rounded rectangle."""
    x, y = xy
    ax.add_patch(patches.FancyBboxPatch(
        (x, y), width, height,
        linewidth=2.5,
        edgecolor=color,
        facecolor=color,
        alpha=alpha,
        boxstyle="round,pad=0.02",
        zorder=3
    ))
    ax.add_patch(patches.FancyBboxPatch(
        (x, y), width, height,
        linewidth=2.5,
        edgecolor=color,
        facecolor="none",
        boxstyle="round,pad=0.02",
        zorder=4
    ))

def draw_matrix(ax, M, title, hl_row=None, hl_col=None):
    m, n = M.shape

    ax.imshow(M, cmap=CMAP, aspect="auto",
              vmin=M.min() - 1, vmax=M.max() + 1)

    # numbers
    for i in range(m):
        for j in range(n):
            ax.text(j, i, f"{int(M[i,j])}",
                    ha="center", va="center",
                    fontsize=13, color=TEXT_COLOR,
                    fontweight="bold")

    # row highlight
    if hl_row is not None:
        add_highlight(ax, (-0.48, hl_row - 0.48), n - 0.04, 0.96, "deepskyblue")

    # column highlight
    if hl_col is not None:
        add_highlight(ax, (hl_col - 0.48, -0.48), 0.96, m - 0.04, "orange")

    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_title(title, fontsize=12, fontweight="bold", pad=8)

# ------------------ PLOT ------------------
fig, axes = plt.subplots(
    1, 3,
    figsize=(13, 4.5),
    gridspec_kw={"width_ratios": [A.shape[1], B.shape[1], C.shape[1]]}
)

draw_matrix(axes[0], A, rf"$A \in \mathbb{{R}}^{{{A.shape[0]} \times {A.shape[1]}}}$", hl_row=hi_row)
draw_matrix(axes[1], B, rf"$B \in \mathbb{{R}}^{{{B.shape[0]} \times {B.shape[1]}}}$", hl_col=hi_col)
draw_matrix(axes[2], C, rf"$C = AB \in \mathbb{{R}}^{{{C.shape[0]} \times {C.shape[1]}}}$")

# highlight result cell
add_highlight(
    axes[2],
    (hi_col - 0.48, hi_row - 0.48),
    0.96, 0.96,
    "limegreen",
    alpha=0.25
)

# ------------------ DOT PRODUCT TEXT ------------------
dot_terms = " + ".join(
    rf"{int(A[hi_row,k])}\cdot{int(B[k,hi_col])}"
    for k in range(A.shape[1])
)

fig.text(
    0.5, -0.04,
    rf"$C_{{{hi_row+1}{hi_col+1}}} = "
    rf"\mathbf{{a}}_{{{hi_row+1}}} \cdot \mathbf{{b}}_{{{hi_col+1}}} "
    rf"= {dot_terms} = {int(C[hi_row, hi_col])}$",
    ha="center",
    fontsize=12,
    color="limegreen",
)

plt.suptitle(
    "Matrix Multiplication: each entry is a dot product",
    fontsize=13,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```
